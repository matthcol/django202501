## Raw SQL
https://docs.djangoproject.com/en/5.1/topics/db/sql/

### raw query without entity result
Use db api2 cursor

```python
sql = """SELECT
    g.genre,
    count(*) as movie_count,
    min(m.year) as first_year,
    max(m.year) as last_year
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %s
GROUP BY g.genre
ORDER BY movie_count DESC
"""
with connection.cursor() as cur:
    cur.execute(sql, ['Clint Eastwood'])
    data = cur.fetchall()
data
```

### raw query with entity result
Use Movie.objects.raw: query must include PK => RawQuerySet

NB: you can add any computed field (not in the model)

```python
reset_queries()
sql = """SELECT
    m.*,
    g.genre as genre
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %s
ORDER BY m.year DESC
"""
rqs = Movie.objects.raw(sql, ['Clint Eastwood'])
rqs
```

```python
prev_title = None
for movie in rqs:
    if movie.title != prev_title:
        print()
    print(movie.genre, ':', movie)  
    prev_title = movie.title
```

## Association

### Fetch associated objects after initial query
Lazy loading: only foreign key director_id is set when each movie is read,
director Person will be set when accessed: new SQL query each time (if not read before)

```python
reset_queries()
data = Movie.objects.filter(year=1984).all()
```

```python
for movie in data:
    print(movie, 'director id:', movie.director_id)
```

```python
# only entities Movie are fetched (no director and actors Person)
connection.queries
```

```python
# when director is accessed, a new query is done ffor each one (SQL: N+1)
for movie in data:
    print(movie)
    print('director:', movie.director)
    print()
```

```python
connection.queries
```

```python
# acessing actors:
# movie.actors is neither a list neither a QuerySet but a ManyRelatedManager
    # you can manage actors with it:
    # - read (.all|filter|first|iterator|...), 
    # - add (.add) 
    # - remove (.remove)
    # - replace all (.set)
# NB: you cannot iterate or slice directly the manager
#     'RelatedManager' object is not iterable
reset_queries()
for movie in data:
    print(movie)
    print('actors:')
    for actor in movie.actors.all():
        print("\t-", actor)
    print()
```

```python
# each access to the actors of one movie generates a new SQL query
# on tables play join person
connection.queries
```

```python
# you can also convert related entities in dictionaries
print(movie.actors.values())
```

### Prefetching
- with +1 query
- with initial query

#### prefetch_related

```python
reset_queries()
data = Movie.objects \
    .filter(year=1984) \
    .prefetch_related('director') \
    .all()
for movie in data:
    print(movie, 'director id:', movie.director_id)
print()
connection.queries
```

```python
# no extra queries when accessing director
reset_queries()
for movie in data:
    print(movie, 'director:', movie.director)
print()
connection.queries
```

```python
# prefetching several objects
```

```python
reset_queries()
data = Movie.objects \
    .filter(year=1984) \
    .prefetch_related('director', 'actors') \
    .all()
for movie in data:
    pass
connection.queries
```

```python
# no extra queries when accessing related objects
reset_queries()
for movie in data:
    print(movie)
    print('\t- director:', movie.director)
    print('\t- actors:')
    for actor in movie.actors.all():
        print('\t\t-', actor)
    print()
connection.queries
```

#### select_related
NB: only available for foreign keys not ManyToMany

```python
reset_queries()
data = Movie.objects \
    .filter(year=1984) \
    .select_related('director') \
    .prefetch_related('actors') \
    .all()
for movie in data:
    pass
connection.queries # only 2 queries
```

```python
reset_queries()
for movie in data:
    print(movie)
    print('\t- director:', movie.director)
    print('\t- actors:')
    for actor in movie.actors.all():
        print('\t\t-', actor)
    print()
connection.queries
```
