## Raw SQL
https://docs.djangoproject.com/en/5.1/topics/db/sql/

```python
sql = """SELECT
    g.genre,
    count(*) as movie_count,
    min(m.year) as first_year,
    max(m.year) as last_year
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %s
GROUP BY g.genre
ORDER BY movie_count DESC
"""
# rqs = Movie.objects.raw(sql, ['Clint Eastwood'])
# rqs (must include PK => <RawQuerySe)
with connection.cursor() as cur:
    cur.execute(sql, ['Clint Eastwood'])
    data = cur.fetchall()
data
```

```python
sql = """SELECT
    g.genre,
    count(*) as movie_count,
    min(m.year) as first_year,
    max(m.year) as last_year
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %(name)s
GROUP BY g.genre
ORDER BY movie_count DESC
"""
with connection.cursor() as cur:
    cur.execute(sql, {'name':'Clint Eastwood'})
    data = cur.fetchall()
data
```

### Navigate from entity Play + fetching

```python
# NB: select_related only for ForeignKey (* to one)
reset_queries()
data = Play.objects \
    .filter(role__icontains='james bond') \
    .select_related('actor', 'movie') \
    .all()    
for play in data:
    print(f"{play.actor.name} plays {play.role} in {play.movie}")
```

### Fetching from entity movie

```python
# NB: to improve ...  .prefetch_related('play_set') \
reset_queries()
movie = Movie.objects \
    .filter(title='The Terminator', year=1984) \
    .select_related('director')  \
    .prefetch_related('actors') \
    .get()
print(movie)
connection.queries
```

```python
reset_queries()
print(movie.director)
for play in movie.play_set.all():
    #print("-", play.actor, ':', play.role)
    print("-", play.actor)
```
