```python
data = Movie.objects \
    .filter(title__istartswith='the ') \
    .order_by('year', 'title')
for m in data[:15]:
    print(m)
```

```python
# order DESC
data = Movie.objects \
    .filter(title__istartswith='the ') \
    .order_by('-year', 'title')
for m in data[:15]:
    print(m)
```

## Raw SQL
https://docs.djangoproject.com/en/5.1/topics/db/sql/

```python
sql = """SELECT
    g.genre,
    count(*) as movie_count,
    min(m.year) as first_year,
    max(m.year) as last_year
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %s
GROUP BY g.genre
ORDER BY movie_count DESC
"""
# rqs = Movie.objects.raw(sql, ['Clint Eastwood'])
# rqs (must include PK => <RawQuerySe)
with connection.cursor() as cur:
    cur.execute(sql, ['Clint Eastwood'])
    data = cur.fetchall()
data
```

```python
sql = """SELECT
    g.genre,
    count(*) as movie_count,
    min(m.year) as first_year,
    max(m.year) as last_year
FROM 
    movie m 
    join person p on m.director_id = p.id
    join have_genre g on g.movie_id = m.id
WHERE p.name = %(name)s
GROUP BY g.genre
ORDER BY movie_count DESC
"""
with connection.cursor() as cur:
    cur.execute(sql, {'name':'Clint Eastwood'})
    data = cur.fetchall()
data
```
